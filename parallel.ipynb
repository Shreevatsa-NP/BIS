```python
from multiprocessing import Pool

def update_chunk(args):
    road, start, end = args
    n = len(road)
    new_chunk = []

    for i in range(start, end):
        left = road[(i - 1) % n]
        curr = road[i]
        right = road[(i + 1) % n]

        # Traffic rule: car moves right if right cell is empty
        if curr == 1 and right == 0:
            new_chunk.append(0)
        elif curr == 0 and left == 1:
            new_chunk.append(1)
        else:
            new_chunk.append(curr)

    return new_chunk

if __name__ == "__main__":
    road = [1, 0, 1, 1, 0, 0, 1, 0, 0, 1, 0, 0]
    workers = 3
    chunk_size = len(road) // workers

    with Pool(workers) as pool:
        for step in range(6):
            print(f"Step {step}: {road}")
            tasks = [(road, i * chunk_size, (i + 1) * chunk_size) for i in range(workers)]
            results = pool.map(update_chunk, tasks)
            road = [cell for chunk in results for cell in chunk]
```
