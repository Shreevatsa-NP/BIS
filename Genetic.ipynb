```python
import random
data = { 0: 18,15: 19,30: 21, 45: 23, 60: 22,  75: 20,  90: 18}
population = [random.choice(list(data.keys())) for _ in range(10)]
for generation in range(20):
    population.sort(key=lambda angle: data[angle], reverse=True)
    best = population[:4]
    population = best.copy()
    while len(population) < 10:
        angle = random.choice(best)
        angle += random.choice([-15, 0, 15])
        angle = max(0, min(90, angle))
        angle = min(data.keys(), key=lambda x: abs(x - angle))
        population.append(angle)
best_angle = max(data, key=data.get)
best_voltage = data[best_angle]
print("Best Angle:", best_angle, "degrees")
print("Highest Voltage:", best_voltage, "Voltage")
```
