```python
import random
books = [
    {"name": "Book A", "weight": 2.0},
    {"name": "Book B", "weight": 3.5},
    {"name": "Book C", "weight": 1.5},
    {"name": "Book D", "weight": 4.0},
    {"name": "Book E", "weight": 2.5},
    {"name": "Book F", "weight": 3.2},
    {"name": "Book G", "weight": 1.5},
    {"name": "Book H", "weight": 2.1}
]
NUM_SHELVES = 3
NUM_PARTICLES = 50
ITERATIONS = 20
W = 0.7
C1 = 1.5
C2 = 1.5
def fitness(position):
    shelf_weights = [0] * NUM_SHELVES
    for i, shelf in enumerate(position):
        shelf_weights[shelf] += books[i]["weight"]
    imbalance = max(shelf_weights) - min(shelf_weights)
    return imbalance
def create_particle():
    return [random.randint(0, NUM_SHELVES - 1)
            for _ in range(len(books))]
particles = []
for _ in range(NUM_PARTICLES):
    position = create_particle()
    particle = {
        "position": position,
        "velocity": [0] * len(books),
        "best_position": position[:],
        "best_fitness": fitness(position)
    }
    particles.append(particle)
global_best_particle = min(
    particles,
    key=lambda p: p["best_fitness"]
)
global_best_position = global_best_particle["best_position"][:]
global_best_fitness = global_best_particle["best_fitness"]

for iteration in range(ITERATIONS):

    for particle in particles:
        for i in range(len(books)):
            r1 = random.random()
            r2 = random.random()
            velocity = (
                W * particle["velocity"][i]
                + C1 * r1 *
                (particle["best_position"][i]
                 - particle["position"][i])
                + C2 * r2 *
                (global_best_position[i]
                 - particle["position"][i])
            )
            particle["velocity"][i] = velocity
            if random.random() < 1 / (1 + abs(velocity)):
                movement = 0
            else:
                movement = 1 if velocity > 0 else -1
            new_shelf = (
                particle["position"][i] + movement
            )
            new_shelf = max(
                0,
                min(NUM_SHELVES - 1, new_shelf)
            )
            particle["position"][i] = new_shelf
        current_fitness = fitness(
            particle["position"]
        )
        if current_fitness < particle["best_fitness"]:
            particle["best_fitness"] = current_fitness
            particle["best_position"] = (
                particle["position"][:]
            )
        if current_fitness < global_best_fitness:
            global_best_fitness = current_fitness
            global_best_position = (
                particle["position"][:]
            )
    print(
        f"Iteration {iteration + 1}: "
        f"Best Fitness = {global_best_fitness:.2f}"
    )
print("\nOptimal Book Arrangement:")
shelf_weights = [0] * NUM_SHELVES
for shelf in range(NUM_SHELVES):
    print(f"\nShelf {shelf + 1}:")
    for i, assigned_shelf in enumerate(global_best_position):
        if assigned_shelf == shelf:
            print(
                f"  {books[i]['name']} "
                f"(Weight: {books[i]['weight']} kg)"
            )
            shelf_weights[shelf] += books[i]["weight"]
print("\nShelf Weights:")
for i, weight in enumerate(shelf_weights):
    print(f"Shelf {i + 1}: {weight:.2f} kg")
print(
    f"\nFinal imbalance: "
    f"{max(shelf_weights) - min(shelf_weights):.2f} kg"
)
```
